```python
df = raw.iloc[5:, [1, 6, 7]].copy()

df.columns = ["date", "rural_cpi", "urban_cpi"]

# Convert date
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Convert CPI values
df["rural_cpi"] = pd.to_numeric(df["rural_cpi"], errors="coerce")
df["urban_cpi"] = pd.to_numeric(df["urban_cpi"], errors="coerce")

# Remove invalid rows
df = df.dropna(subset=["date", "rural_cpi", "urban_cpi"])

# Sort chronologically
df = df.sort_values("date").reset_index(drop=True)

print(df.head())
print()
print(df.tail())
print()
print(df.dtypes)
```

```python
# Year-on-year inflation
df["rural_inflation"] = df["rural_cpi"].pct_change(12) * 100
df["urban_inflation"] = df["urban_cpi"].pct_change(12) * 100

# Rural minus Urban inflation
df["gap"] = df["rural_inflation"] - df["urban_inflation"]

# Keep the period relevant to the story
story_df = df[
    (df["date"] >= "2024-01-01") &
    (df["date"] <= "2026-07-31")
].copy()

# Show the results
print(
    story_df[
        ["date", "rural_inflation", "urban_inflation", "gap"]
    ].to_string(index=False)
)
```

```python
print("Average rural inflation:",
      round(story_df["rural_inflation"].mean(), 2))

print("Average urban inflation:",
      round(story_df["urban_inflation"].mean(), 2))

print("Average rural-urban gap:",
      round(story_df["gap"].mean(), 2))

print("Months rural inflation > urban:",
      (story_df["gap"] > 0).sum())

print("Months urban inflation > rural:",
      (story_df["gap"] < 0).sum())

print("Largest rural premium:")
print(
    story_df.loc[
        story_df["gap"].idxmax(),
        ["date", "rural_inflation", "urban_inflation", "gap"]
    ]
)

print("Largest urban premium:")
print(
    story_df.loc[
        story_df["gap"].idxmin(),
        ["date", "rural_inflation", "urban_inflation", "gap"]
    ]
)
```

```python
# Classify which side has higher inflation
story_df["winner"] = story_df["gap"].apply(
    lambda x: "Rural higher" if x > 0
    else "Urban higher" if x < 0
    else "Equal"
)

# Identify changes in direction
story_df["change"] = (
    story_df["winner"] != story_df["winner"].shift()
)

print(
    story_df[
        ["date", "rural_inflation", "urban_inflation", "gap", "winner", "change"]
    ].to_string(index=False)
)
```

```python
clean_df = story_df[
    ["date", "rural_inflation", "urban_inflation", "gap"]
].copy()

clean_df["date"] = clean_df["date"].dt.strftime("%b %Y")

clean_df["rural_inflation"] = clean_df["rural_inflation"].round(2)
clean_df["urban_inflation"] = clean_df["urban_inflation"].round(2)
clean_df["gap"] = clean_df["gap"].round(2)

print(clean_df.to_string(index=False))
```

```python
annual = story_df.groupby(
    story_df["date"].dt.year
).agg(
    avg_rural_inflation=("rural_inflation", "mean"),
    avg_urban_inflation=("urban_inflation", "mean"),
    avg_gap=("gap", "mean"),
    max_gap=("gap", "max"),
    min_gap=("gap", "min")
)

annual.round(2)
```

```python
story_df["rural_higher"] = story_df["gap"] > 0

story_df["switch"] = (
    story_df["rural_higher"] !=
    story_df["rural_higher"].shift(1)
)

switches = story_df[
    story_df["switch"]
][[
    "date",
    "rural_inflation",
    "urban_inflation",
    "gap"
]]

switches.round(2)
```

```python
story_df["direction"] = story_df["gap"].apply(
    lambda x: "Rural higher" if x > 0 else "Urban higher"
)

story_df["period"] = (
    story_df["direction"] != story_df["direction"].shift()
).cumsum()

episodes = story_df.groupby("period").agg(
    direction=("direction", "first"),
    start=("date", "min"),
    end=("date", "max"),
    months=("date", "count"),
    avg_gap=("gap", "mean"),
    max_gap=("gap", "max"),
    min_gap=("gap", "min")
)

episodes.round(2)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

recent = df[
    (df["date"] >= "2024-01-01") &
    (df["date"] <= "2026-07-31")
].copy()

plt.figure(figsize=(14, 7))

plt.plot(
    recent["date"],
    recent["gap"],
    linewidth=2.5
)

# Zero line
plt.axhline(0, linewidth=1)

# Shade rural/urban lead
plt.fill_between(
    recent["date"],
    recent["gap"],
    0,
    where=recent["gap"] > 0,
    alpha=0.12
)

plt.fill_between(
    recent["date"],
    recent["gap"],
    0,
    where=recent["gap"] < 0,
    alpha=0.08
)

# Labels
plt.title(
    "India's Rural–Urban Inflation Gap Has Reversed",
    fontsize=20,
    fontweight="bold",
    loc="left",
    pad=20
)

plt.text(
    0, 1.02,
    "Rural CPI inflation minus urban CPI inflation, percentage points",
    transform=plt.gca().transAxes,
    fontsize=11
)

plt.ylabel("Percentage points")
plt.xlabel("")

# Date formatting
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))

plt.xticks(rotation=45)

# Clean look
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()
plt.show()
```

```python
annual = (
    recent
    .groupby("year")
    .agg(
        rural=("rural_inflation", "mean"),
        urban=("urban_inflation", "mean"),
        gap=("gap", "mean")
    )
    .reset_index()
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    annual["year"].astype(str),
    annual["gap"],
    width=0.6
)

plt.axhline(0, linewidth=1)

plt.title(
    "The Rural–Urban Inflation Gap Has Changed Direction",
    fontsize=19,
    fontweight="bold",
    loc="left",
    pad=20
)

plt.text(
    0, 1.02,
    "Average rural inflation minus urban inflation, percentage points",
    transform=plt.gca().transAxes,
    fontsize=10.5
)

plt.ylabel("Percentage points")

ax = plt.gca()

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.grid(
    axis="y",
    alpha=0.2
)

# Add values above/below bars
for bar, value in zip(bars, annual["gap"]):
    if value >= 0:
        y = value + 0.04
        va = "bottom"
    else:
        y = value - 0.04
        va = "top"

    plt.text(
        bar.get_x() + bar.get_width()/2,
        y,
        f"{value:.2f}",
        ha="center",
        va=va,
        fontsize=10
    )

plt.tight_layout()
plt.show()
```
