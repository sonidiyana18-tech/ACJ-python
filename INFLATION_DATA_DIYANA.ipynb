# Now, I will try to find the state with the highest inflation rate by sorting values according to "inflation combined" and in ascending order (top 10)

```python
df.sort_values("Inflation Combined (%)", ascending=False).head(10)
```

### Telangana: 6.32% 
# Now, the state with the lowest inflation rate

```python
df.sort_values("Inflation Combined (%)", ascending=True).head(10)
```

### Mizoram: 1.84%
### Telangana: 6.32% vs. Mizoram: 1.84%
### That's a difference of 4.48 percentage points.

# I will try to do a rural vs. urban inflation data analysis

```python
df.sort_values(
    by="Inflation Rural (%)",
    ascending=False
)[["State/UT", "Inflation Rural (%)"]].head(10)
```

```python
df.sort_values(
    by="Inflation Rural (%)",
    ascending=False
)[["State/UT", "Inflation Urban (%)"]].head(10)
```

# Finding where rural inflation is higher than urban and by how much

```python
df["Rural-Urban Inflation Gap"] = (
    df["Inflation Rural (%)"] - df["Inflation Urban (%)"]
)
```

```python
df[["State/UT", "Rural-Urban Inflation Gap"]].sort_values(
    "Rural-Urban Inflation Gap", ascending=False
).head(10)
```

### In Andaman and Nicobar Islands rural inflation is 1.73 percentage points higher than urban inflation

# Finding where urban inflation is higher than rural

```python
df[["State/UT", "Rural-Urban Inflation Gap"]].sort_values(
    "Rural-Urban Inflation Gap", ascending=True
).head(10)
```

### Himachal Pradesh has urban inflation surpassing rural by 1.18 percentage points

# Calculate the average rural vs urban inflation

```python
float(round(df["Inflation Rural (%)"].mean(), 2))
```

```python
float(round(df["Inflation Urban (%)"].mean(), 2))
```

### Average rural inflation: 4.27%

### Average urban inflation: 3.62%

# Finding how many states have higher rural inflation

```python
float((df["Inflation Rural (%)"] > df["Inflation Urban (%)"]).sum())
```

# Finding how many states have higher urban inflation

```python
float((df["Inflation Urban (%)"] > df["Inflation Rural (%)"]).sum())
```

# Rural India Faces Higher Inflation Than Urban Areas

### Story pitch

**What:**
Rural inflation is higher than urban inflation across several Indian states, with Telangana recording the highest combined inflation rate at 6.32%, while Mizoram recorded the lowest at 1.84%.

**Why it matters:**
Inflation does not affect all Indians equally. The gap between rural and urban inflation suggests that households in different parts of the country are facing different price pressures. Understanding where rural inflation is higher — and where urban inflation is higher — can show how uneven the cost-of-living squeeze is across India.

> **The story will look beyond India's overall inflation rate to show how the inflation experience differs between rural and urban households across states.**
